```python
# Define a tool function
@tool
def fetch_weather(city: str, date: str):
    """Simulate fetching weather data"""
    return f"The weather in {city} on {date} is sunny with a high of 75°F."
```

```python
# Bind the tool to the model
llm = chatllm.bind_tools([fetch_weather])
```

```python
# User's input
messages = [
    {"role": "user", "content": "What will the weather be in San Francisco tomorrow?"}
]
```

```python
# Model processes the input and invokes the tool
result = llm.invoke(messages)

# Output the response
print(result.tool_calls)
```
